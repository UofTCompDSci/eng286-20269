## Key pandas features

Here are the pandas features you will need. All of them have come up in the Week 7 pandas notebooks or the Week 8 tutorial.

- `pd.read_csv()`: read a CSV file into a DataFrame
- `len()`: count the rows of a DataFrame
- `df[["col1", "col2"]]` and `.head()`: look at some of the columns, and the first few rows
- `.replace({...})`: change the values in a column using a dictionary
- `df["new column"] = ...`: add a new column
- `.value_counts()`: count how many times each value appears in a column
- `df[df["col"] == value]`: keep only the rows that match
- `.value_counts(normalize=True)`: get each value's share of the rows, instead of its count
- `.round()`: round every number in a Series

Before you start this task, here is an example of a very useful pandas method.  `.replace()` takes the values in a Series, looks up each value in a dictionary, and replaces the ones that it finds keys for with the values in the dictionary.
