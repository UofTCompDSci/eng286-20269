## Key pandas features

Here are the pandas features we'll use. Some will be familiar and some will be new. Each one is explained in the task where you first need it.

**Loading and looking at a table**
- `pd.read_csv()`: read a CSV file into a DataFrame
- `df[["col1", "col2"]]`: keep only some of the columns
- `.head()`: look at the first few rows

**Counting**
- `.value_counts()`: count how many times each value appears in a column
- `.value_counts(normalize=True)`: get each value's share of the rows, instead of its count
- `.round()`: round every number in a Series

**Filtering rows**
- `df[df["col"] == value]` and `df[df["col"] != value]`: keep the rows that match, or that don't match
- `.isin([...])`: check whether each value is in a list

**Changing a table**
- `df["new column"] = ...`: add a new column
- `.str.split()` and `.str[1]`: split every string in a column, and take one piece from each
- `.replace({...})`: fix values using a dictionary of corrections

**Combining and charting**
- `pd.DataFrame({...})`: build a table from two Series
- `.fillna(0)`: fill in missing values
- `.sort_values()`: put the rows in order
- `.plot.barh()`: draw a horizontal bar chart

#### Part B: From counts to percentages

Counts are hard to compare when one bar is so much bigger than the rest. Another way to describe the data is with **percentages**.

`.value_counts(normalize=True)` gives the *fraction* of rows for each value instead of the count. Multiplying by 100 turns the fractions into percentages, and `.round(1)` keeps one decimal place. For example:

```python
status_percent = books["Status"].value_counts(normalize=True) * 100
status_percent = status_percent.round(1)
```

Notice that we didn't need a loop to multiply by 100: pandas does the calculation for every value in the Series.

### Task 4: Which Publishers?

#### Part A: Cleaning the publisher names

Count the `Publisher` column, and look closely at the results. You should find the same publisher written in two different ways: `Douglas & McIntyre` and `D & M Publishers` (the short form of the same company's name). If we don't fix this, the publisher's books are split across two rows, and its count is too low.

Two other names are very long: `Esplanade Books, the fiction imprint at Véhicule Press` and `Patrick Crean Editions, an imprint of HarperCollins Publishers`. An *imprint* is a brand name that a larger publisher uses for some of its books.

We can fix all three at once with a **corrections dictionary**: each key is a value to replace, and each value is the name to use instead. `.replace()` applies the whole dictionary to a column. For example:

```python
fixes = {"Random House Canada": "Random House"}
new_names = books["Publisher"].replace(fixes)
```

1. Make a dictionary called `corrections` that makes these three changes:
    - `D & M Publishers` becomes `Douglas & McIntyre`
    - `Esplanade Books, the fiction imprint at Véhicule Press` becomes `Véhicule Press`
    - `Patrick Crean Editions, an imprint of HarperCollins Publishers` becomes `HarperCollins Canada`
2. Use `.replace()` to fix the `Publisher` column of `books`.
3. Count the cleaned `Publisher` column, and store the result in a variable called `publisher_counts`.
4. Store the number of different publishers in a variable called `number_of_publishers`. (Hint: `publisher_counts` has one value for each publisher.)

When you're done, `number_of_publishers` should be `46`, and `publisher_counts["HarperCollins Canada"]` should be `47`.

````python
# Solution
# Write your code here

print(books["Publisher"].value_counts())

corrections = {
    "D & M Publishers": "Douglas & McIntyre",
    "Esplanade Books, the fiction imprint at Véhicule Press": "Véhicule Press",
    "Patrick Crean Editions, an imprint of HarperCollins Publishers": "HarperCollins Canada",
}

books["Publisher"] = books["Publisher"].replace(corrections)

publisher_counts = books["Publisher"].value_counts()

number_of_publishers = len(publisher_counts)

# Check your work
print(number_of_publishers)
print(publisher_counts["HarperCollins Canada"])
````

#### Part D: Nominations and wins side by side 

Let's put nominations and wins in one chart. To do that, we need one table with a column for each.

`pd.DataFrame` builds a table from Series. We give it a dictionary, where each key is a column name and each value is a Series:

```python
new_table = pd.DataFrame({"First column": first_series, "Second column": second_series})
```

pandas matches up the two Series by publisher name. There's a catch, though: a publisher with no wins isn't in `winner_counts` at all, so its `Won` value is missing. pandas shows a missing value as `NaN`, and `.fillna(0)` replaces each one with a `0`.

Also, the new table is in alphabetical order. `.sort_values("Column name", ascending=False)` puts the rows in order from biggest to smallest, using the column we name.

1. Make a DataFrame called `by_publisher` with two columns: `Nominated` (from `publisher_counts`) and `Won` (from `winner_counts`). Use `.fillna(0)` to fill in the missing values.
2. Sort `by_publisher` from most to fewest nominations.
3. Draw a horizontal bar chart of the first 12 rows. Use the same steps as before, but give `color` a list of two colours, one for each column: `color=["#2a78d6", "#eb6834"]`.


````python
# Provided solution
by_publisher = pd.DataFrame({"Nominated": publisher_counts, "Won": winner_counts}).fillna(0)
by_publisher = by_publisher.sort_values("Nominated", ascending=False)

# Check your work
print(by_publisher.head())

ax = by_publisher.head(12).plot.barh(color=["#2a78d6", "#eb6834"])

ax.invert_yaxis()
ax.set_title("McClelland & Stewart wins most often, but isn't nominated most")
ax.set_xlabel("Number of books")
ax.set_ylabel("")
plt.show()
````
