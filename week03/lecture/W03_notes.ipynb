## A. List of phonemes

You can find the CMU list of 39 phonemes on their [web site](http://www.speech.cs.cmu.edu/cgi-bin/cmudict?in=poem&stress=-s#phones). These are the notations for the different sounds in the language. We will mostly explore them by example.



````python
# Run this cell but don't change this code.
# You are not expected to understand the code in this cell just yet, 
# but I've added comments to help you read it.

# The first three lines import the nltk library and set up to use the cmudictionary
import nltk
nltk.download('cmudict')
from nltk.corpus import cmudict

def get_pronunciation(word):
    ''' Return the first phoneme list found for the word word
    '''
    # Create a Python dictionary from the cmudict and store it in pronounce_dictionary
    pronounce_dictionary = cmudict.dict()

    # Ensure the word we want to look up is in lower case
    word = word.lower()

    # We check first to make sure that word is in the dictionary to 
    # avoid getting a nasty error message
    if word in pronounce_dictionary:

        # If the word is in the dictionary then return its first pronunciation list
        return pronounce_dictionary[word][0]
    else:
        # If it is not in the dictionary return the value None
        return None
````

````python
# Now let's lookup a handful of words to see what their pronunciation looks like
# Remember to run the above cell that defines get_pronunciation before this one
# What type is returned by get_pronunciation()?

print(get_pronunciation("poem"))

print(get_pronunciation("cat"))

print(get_pronunciation("elephant"))

print(get_pronunciation("I"))

print(get_pronunciation("a"))

print(get_pronunciation("the"))

print(get_pronunciation("stout"))

print(get_pronunciation("while"))

print(get_pronunciation("dictionary"))

````

## 2. String methods
Before we can decide whether a phoneme represents a syllable we need a way to find out whether a string or character is a number. To do that we need to talk about _methods_.



*Methods*, which are a special kind of function that only belong to certain data types and which are written out differently than functions.

As you'll recall, the syntax of a function is `function(argument)`, which we said was analogous to `verb(noun)`: functions are verb-y in that they *perform actions* on noun-like arguments.
But not all kinds of actions are applicable or appropriate to all kind of nouns, as you will perhaps have discovered in your madlibs exercises last week (Can you really be "eaten alive" by an "uproarious smoke"?).

**And not all functions are appropriate to all kinds of Python data types**.

It makes sense to `print()` anything or check the `type()` of anything — but whereas it certainly makes sense to **convert a `str` to all lowercase letters**, it makes absolutely none to convert an `int` or a `float` or a `bool` to lowercase letters, since they are not composed of letters.
This is not a hypothetical example: Python has a built-in way to turn strings into lowercase letters. But because it is something that only makes sense for one data type (`str`s), it is a **method** rather than a **function**.

The syntax for methods is as follows: 

> `data.method(argument)`

As you can see, they look quite a bit like functions, but they come **after** the data that you want to perform some action on, and they are "attached" to that data, as it were, by a period (`.`). 
Think of methods like **suffixes** — or, if you speak an inflected language like French or Polish, like grammatical words endings like the bolded bits in "Tu aim**es**" vs. "Nous aim**ons**."

Python has lots of cool string-specific methods. Let's explore a few:

* `.replace()`: replace some text with other text (like "Find and Replace" in Word)
* `.isdigit()`: returns true if the string is all numbers, false otherwise

### `if` statements

An `if` statement is an instruction to do something *if* a particular condition is met.

```
if condition is true:
    do something
```


Here's a Python `if` statement:

````python
distance = 63
if distance > 50:
    print("That's an impressive bike ride")
````

````python
city = "Toronto"
if len(city) > 5:
    print(f"{city} is a long city name.")

if city == "Ottawa":
    print(f"This city is the capital of Canada.")
````

Now let's apply this to our question of the number of syllables in a word.

The number of syllables in a word is the same as the number of phonemes that end with a digit indicating the stress of that vowel.

We could say it like this:

"For each phoneme in the phoneme list, if it ends with a digit, add one to the count of syllables."

````python
word = "elephant"
word_p = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

count = 0

for phoneme in word_p:
    last_character = phoneme[-1]
    if last_character.isdigit():
        count = count + 1
print(f"The number of syllables in {word} is {str(count)}")

````

## Another approach

Instead of incrementing a count variable, we could build up a string of the stress values, and then get the length of the string.

````python
word = "elephant"
word_p = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

stress = "" # empty string to start

for phoneme in word_p:
    last_character = phoneme[-1]
    if last_character.isdigit():
        stress = stress + last_character
print(f"The number of syllables in {word} is {str(len(stress))}")
print(f"The stress pattern of {word} is {stress}")
````
