Pronounciation Methods:

- International Phonetic Alphabet (IPA)
    - used widely
    - challenging to learn

- CMU Dictionary 
    - all characters are regular English characters
    - useful for processing
    - used in the Natural Langauge Toolkit (nltk), a Python library

When we look up a word in the CMU dictionary, we will get a list of phonemes.  That's the first step.  You can read the list of 39 phonemes on their [web site](http://www.speech.cs.cmu.edu/cgi-bin/cmudict?in=poem&stress=-s#phones).


````python
# Run this cell but don't change this code.

import nltk
nltk.download('cmudict')
from nltk.corpus import cmudict

def get_pronunciation(word):
    ''' Return the first phoneme list found for the word word
    '''
    pronounce_dictionary = cmudict.dict()
    word = word.lower()

    if word in pronounce_dictionary:
        return pronounce_dictionary[word][0]
    else:
        return None
````

## 4. Looping over the elements of a list

Before we talk about loops, let's introduce a useful function: `len()` which returns the length of a collection (or sequence).

````python
# print the length of the list stored in word_p
# Notice that this gives us the number of phonemes in the word
print(len(word_p))

# print the length of the string stored in word_p[0]
# This gives us the number of characters in the string stored in word_p[0]
print(len(word_p[0]))
````

A _for loop_ allows us to move through the elements of a particular list variable -- iterate over it -- and do something to each element of the list.

For example, to ask someone to prepare some cut up vegetables you might say:

```
for carrot in bag_of_carrots:
    cut_ends(carrot)
    peel(carrot)
```

Notice that the variable `carrot` takes on the value of the next item from the `bag_of_carrots` each time through the loop.  So we perform these operations on each carrot.

Moving away from analogies, let's print each element of a list:


````python
for phoneme in word_p:
    print(phoneme)
````

We will add one more piece to this loop which is to count the number of elements of the list and print the count when the loop completes. To do this we need a variable `count` that we will increment on each iteration of the loop.

````python
count = 0
for phoneme in word_p:
    print(phoneme)
    count = count + 1  # Add 1 to count

print(f"There are {count} phonemes in {word_p}.")
````

Now let's apply this to our question of the number of syllables in a word.

"For each phoneme in the phoneme list, if it ends with a digit, add one to the count of syllables."

````python
word = "elephant"
word_p = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

count = 0

for phoneme in word_p:
    last_character = phoneme[-1]
    if last_character.isdigit():
        count = count + 1
        
print(f"The number of syllables in {word} is {str(count)}")

````

## Another approach

Instead of incrementing a count variable, we could build up a string of the stress values, and then get the length of the string.

````python
word = "elephant"
word_p = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

stress = "" # empty string to start

for phoneme in word_p:
    last_character = phoneme[-1]
    if last_character.isdigit():
        stress = stress + last_character
        
print(f"The number of syllables in {word} is {str(len(stress))}")
print(f"The stress pattern of {word} is {stress}")
````
