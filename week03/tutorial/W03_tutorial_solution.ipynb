### Task 4: Loops + If Statements — Counting Long Words

Now let's combine everything: loop over every word in a list, check a condition with an `if` statement, and keep a running count. This is exactly the pattern lecture used to count syllables — loop, check a condition, increment a counter — just applied to a new question.

Next week we will use a similar approach to count the number of syllables in each word.

Here are the last words of each line in Shakespeare's Sonnet 18:

```python
words = ['day', 'temperate', 'May', 'date', 'shines', 'dimmed', 'declines',
         'untrimmed', 'fade', 'ownest', 'shade', 'growest', 'see', 'thee']
```

1. Write a `for` loop that goes through `words` and counts how many are "long" (more than 4 letters). Store the count in a variable called `long_count`.
2. Print `long_count`. Verify it by counting the number of long words by hand.



````python
# Write your code here
words = ['day', 'temperate', 'May', 'date', 'shines', 'dimmed', 'declines',
         'untrimmed', 'fade', 'ownest', 'shade', 'growest', 'see', 'thee']


````

````python
# Solution

words = ['day', 'temperate', 'May', 'date', 'shines', 'dimmed', 'declines',
         'untrimmed', 'fade', 'ownest', 'shade', 'growest', 'see', 'thee']

# Count the long words
long_count = 0
for word in words:
    if len(word) > 4:
        long_count = long_count + 1
print(f"Number of long words: {long_count}")
````

## Extra Practice 1

Try using your `if` statement from Task 3 to count how many words fit in each of the three categories of short, medium and long words.  The goal is to do this in one loop.

Create 3 variables: `short_count`, `medium_count`, and `long_count`

````python
# Write your code here
words = ['day', 'temperate', 'May', 'date', 'shines', 'dimmed', 'declines',
         'untrimmed', 'fade', 'ownest', 'shade', 'growest', 'see', 'thee']
````

````python
# Solution
words = ['day', 'temperate', 'May', 'date', 'shines', 'dimmed', 'declines',
         'untrimmed', 'fade', 'ownest', 'shade', 'growest', 'see', 'thee']

short_count = 0
medium_count = 0
long_count = 0

for word in words:
    if len(word) <= 3:
        short_count = short_count + 1
    elif len(word) <= 7:
        medium_count = medium_count + 1
    else:
        long_count = long_count + 1
print(f"Number of short words: {short_count}")
print(f"Number of medium words: {medium_count}")
print(f"Number of long words: {long_count}")
````

Finally, try putting this together with the `.split()`method to count the lengths of the words in the whole sonnet.  Note that our current method counts punctuation as part of the adjacent word.  We will learn how to deal with this in a subsequent week.

Here is one more new concept for you. We want to fit the whole sonnet into a single string variable, but we want to preserve the line breaks. To do this we use a special character `\n` which is called the _newline_ character. (It looks like two characters, but is treated as a single character.)

What is really helpful is that `\n` is considered a _white space_ character like a _space_ (or blank). The other main whitespace character we might come across is a _tab_ (`\t`). The reason this is helpful is that the default behaviour of `.split()` is to split on white space, so the newline characters are treated as white space separators and the result is what we want.



````python
# Here is a string with newline characters in it

sonnet = "Let me not to the marriage of true minds\nAdmit impediments; love is not love\nWhich alters when it alteration finds,\nOr bends with the remover to remove. \nO no, it is an ever-fixèd mark\nThat looks on tempests and is never shaken;\nIt is the star to every wand'ring bark\nWhose worth's unknown, although his height be taken.\nLove's not time's fool, though rosy lips and cheeks\nWithin his bending sickle's compass come.\nLove alters not with his brief hours and weeks,\n But bears it out even to the edge of doom:\nIf this be error and upon me proved,\nI never writ, nor no man ever loved."

print(sonnet)

# split will still do what we want!
sonnet_words = sonnet.split()

# Take a look at what is in the list sonnet_words.
print(sonnet_words)
````

````python
# Solution
# Now try counting words of different lengths

short_count = 0
medium_count = 0
long_count = 0

for word in sonnet_words:
    if len(word) <= 3:
        short_count = short_count + 1
    elif len(word) <= 7:
        medium_count = medium_count + 1
    else:
        long_count = long_count + 1
print(f"Number of short words: {short_count}")
print(f"Number of medium words: {medium_count}")
print(f"Number of long words: {long_count}")
````
