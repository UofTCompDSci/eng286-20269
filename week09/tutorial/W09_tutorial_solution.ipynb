### Task 4: Tracing the Counting Pattern

Here is the pattern, counting how many times each author appears in a list:

```python
authors = ["Atwood", "Munro", "Atwood", "Shields", "Munro", "Atwood"]

counts = {}
for author in authors:
    if author in counts:
        counts[author] = counts[author] + 1
    else:
        counts[author] = 1
print(counts)
```

#### Part A

Before you run it, trace the code. On paper (or in the comment), write down what `counts` holds after each time through the loop. The first two times through are done for you:

| Time through the loop | `author` | `counts` after this time through |
|---|---|---|
| 1 | `"Atwood"` | `{"Atwood": 1}` |
| 2 | `"Munro"` | `{"Atwood": 1, "Munro": 1}` |
| 3 | | |
| 4 | | |
| 5 | | |
| 6 | | |

What does the code print?

````python
# Solution
# {'Atwood': 3, 'Munro': 2, 'Shields': 1}
#
# 3: "Atwood"  -> {'Atwood': 2, 'Munro': 1}
# 4: "Shields" -> {'Atwood': 2, 'Munro': 1, 'Shields': 1}
# 5: "Munro"   -> {'Atwood': 2, 'Munro': 2, 'Shields': 1}
# 6: "Atwood"  -> {'Atwood': 3, 'Munro': 2, 'Shields': 1}

authors = ["Atwood", "Munro", "Atwood", "Shields", "Munro", "Atwood"]

counts = {}
for author in authors:
    if author in counts:
        counts[author] = counts[author] + 1
    else:
        counts[author] = 1
print(counts)
````
