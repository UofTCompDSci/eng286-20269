Here is the pattern, counting how many times each author appears in a list:

```python

    if author in counts:
        counts[author] = counts[author] + 1
    else:
        counts[author] = 1

```

### Task 5: Finding the Biggest

In Week 6 we found the most-nominated author with a loop like this one. This is another common pattern.

#### Part A

````python
# My prediction:
#

counts = {"Atwood": 3, "Munro": 2, "Shields": 1}

most = 0
top_author = ""
for author in counts:
    if counts[author] > most:
        most = counts[author]
        top_author = author
print(top_author, most)
````

#### Part B

Now Munro and Atwood are tied, and Munro comes first in the dictionary. Which author is printed?
