# Week 5

## 5.1 Reading a file line by line  (This will not be on the test)

You won't have to write code that opens or reads a file on the test. But you do need to know what it does.  In particular, what the variables `lines` in the example below holds.

To read a file, we **open** it in a `with` block. The `with` block makes sure the file is closed again when we're done with it. `encoding="utf-8"` tells Python how the characters in the file are stored.

`f.readlines()` reads the whole file and returns a **list of strings**, one string for each line of the file. Each line still ends with a newline character `\n`, so we usually use `.strip()` to remove it.

```python
with open("giller_nominees_clean.csv", encoding="utf-8") as f:
    lines = f.readlines()

print(len(lines), "lines in the file")
print(lines[0])           # the header row; the newline at the end prints as a blank line
print(lines[0].strip())   # .strip() removes the newline
```

## 5.2 CSV files and `csv.reader`: a list of lists

A **CSV** (comma-separated values) file stores a table as plain text. Each line is a row, and commas separate the columns. The first line is usually a **header row** with the column names.

We can't just use `.split(",")` on each line, because some values contain commas themselves (like `"Toronto, ON"`). The `csv` module handles this for us. `csv.reader` gives us each row as a list, so the whole table becomes a **list of lists**:
- `rows[0]` is the header row.
- `rows[1]` is the first book, and `rows[1][2]` is the item at index 2 of that row (the title).

You won't have to write the `with open(...)` code on the test. But you should know that **`csv.reader` gives a list of lists**, and how to get a value out of it. For example, `rows` is a list of lists.

```python
import csv

with open("giller_nominees_clean.csv", encoding="utf-8") as f:
    reader = csv.reader(f)
    rows = list(reader)   # turn all of the rows into one big list

print(len(rows), "rows, including the header")
print(rows[0])
print(rows[1][2], "by", rows[1][3])
```

## 5.3 Dictionaries: keys and values

A **dictionary** stores pairs of values: each **key** is linked to a **value**. Dictionaries are written with curly braces `{}`, with a colon between each key and its value.

We look up a value by putting its **key** in square brackets. Looking up a key that isn't in the dictionary causes a `KeyError`. We can only look things up by key, not by value.

```python
winners = {
    "The Book of Secrets": "M.G. Vassanji",
    "A Fine Balance": "Rohinton Mistry",
    "Alias Grace": "Margaret Atwood",
}

print(winners["Alias Grace"])
print(len(winners), "entries")    # len() counts the key-value pairs

# print(winners["Margaret Atwood"])    # What goes wrong? Remove the # to check.
```

## 5.4 `in` with a dictionary

`key in dictionary` is `True` if the key is in the dictionary. It checks the **keys only**, not the values. We often use it to avoid a `KeyError`.

```python
print("Alias Grace" in winners)
print("Margaret Atwood" in winners)   # a value, not a key

title = "Fall on Your Knees"
if title in winners:
    print(f"{title} was written by {winners[title]}")
else:
    print(f"{title} is not in our dictionary")
```

## 5.5 Adding and changing values

Assigning to a key with `=` does one of two things:
- If the key is **not** in the dictionary yet, it **adds** a new key-value pair.
- If the key is already in the dictionary, it **replaces** the old value. A dictionary can't have the same key twice.

Like `.append()` for lists, this changes the original dictionary.

```python
winners["How to Pronounce Knife"] = "Souvankham Thammavongsa"   # a new key
print(len(winners))

winners["Alias Grace"] = "Atwood"   # a key that is already there
print(len(winners))                 # still the same number of keys
print(winners["Alias Grace"])
```

## 5.6 Looping over a dictionary

A `for` loop over a dictionary gives us its **keys**, one at a time. To get each value, we look it up with the key.

```python
for title in winners:
    print(f"{title} ({winners[title]})")
```

## 5.7 `csv.DictReader`: a list of dictionaries

`csv.DictReader` gives us each row as a **dictionary**. The column names from the header row become the keys. So the whole table becomes a **list of dictionaries**, one dictionary for each book.

The header row is used for the keys, so it is not a row of its own. That's why `nominees` has one fewer item than `rows`.

To get one value, we first choose a row with a number, then a column with a key: `nominees[0]["Title"]`.

As with `csv.reader`, you won't have to write the code that reads the file. But you should know that **`csv.DictReader` gives a list of dictionaries**, how that differs from the list of lists `csv.reader` gives, and how to get a value out of each one.

| | `csv.reader` | `csv.DictReader` |
|---|---|---|
| The table is a... | list of lists | list of dictionaries |
| One row is a... | list | dictionary |
| The header row is... | the first row, `rows[0]` | used for the keys, not a row of its own |
| The title of the first book is... | `rows[1][2]` | `nominees[0]["Title"]` |

```python
with open("giller_nominees_clean.csv", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    nominees = list(reader)

print(len(nominees), "nominated books")

book = nominees[0]
print(book["Title"], "by", book["Author"])
print(nominees[1]["Title"])

# print(nominees["Title"])    # What goes wrong? Remove the # to check.
```

## 5.8 Values from a CSV file are strings

A CSV file is just text. So the `csv` module gives us every value as a **string**, even when it looks like a number. The string `"1996"` and the integer `1996` are not equal.

If we need to do arithmetic with a value, we convert it to an integer first with `int()`.

```python
year = nominees[0]["Year"]

print(year)
print(type(year))
print(year == 1996)     # a string is never equal to an integer
print(year == "1996")
print(int(year) + 10)   # int() turns "1996" into 1996

# print(year + 10)      # What goes wrong? Remove the # to check.
```

## 5.9 Choosing rows with a loop and `if`

To find the rows we're interested in, we loop over the table and use an `if` statement to check each row. Here we print every book nominated in 1996. Notice the quotation marks around `"1996"`.

```python
for book in nominees:
    if book["Year"] == "1996":
        print(f"{book['Title']} by {book['Author']}")
```

## 5.10 The counting pattern

To count how many times each value appears, we use a dictionary:
1. **Before** the loop, create an empty dictionary.
2. **Inside** the loop, check whether we've seen this value before:
    - If it's already a key, add 1 to its count.
    - If not, this is the first time we've seen it, so set its count to 1.
3. **After** the loop, the dictionary holds a count for every value.

This is probably the most important pattern from Weeks 5–7. It is so important that in pandas (week 8) it is implemented by the `.value_counts()` method.

```python
nomination_counts = {}

for book in nominees:
    author = book["Author"]
    if author in nomination_counts:
        nomination_counts[author] = nomination_counts[author] + 1
    else:
        nomination_counts[author] = 1

print(len(nomination_counts), "different authors")
print(nomination_counts["Margaret Atwood"])
print(nomination_counts["Alice Munro"])
```

## 6.1 Cleaning values with `.strip()`

Real data is messy. In Week 6, many values in the original Giller file had extra spaces at the start or end, so `"Toronto, ON "` and `"Toronto, ON"` were counted as different places.

- `.strip()` removes spaces and newlines from both ends of a string, but not from the middle.
- `.strip(",")` removes commas from both ends instead.

To clean every value in a row, we build a new list with `.append()`.

```python
messy_row = ["2011 ", " The Antagonist", "Lynn Coady", "Toronto, ON  "]

clean_row = []
for value in messy_row:
    clean_row.append(value.strip())

print(messy_row)
print(clean_row)
print("Lagos, Nigeria,".strip(","))
```

## 6.2 Splitting a value into parts

`.split(",")` breaks a string into a list wherever there is a comma. The pieces keep any spaces that came after the commas, so we usually `.strip()` each piece too.

The last piece, `[-1]`, is the province, state, or country. That works whether a place has one, two, or three parts.

```python
born = "Oak Park, IL, USA"
parts = born.split(",")

print(parts)
print(parts[-1])           # notice the space at the start
print(parts[-1].strip())

print("Montreal, QC and Paris, France".split(" and "))   # we can split on any string
```

## 6.3 Writing a CSV file  (This will not be on the test)

To save a table, we open a file with `"w"` (for **write**) and use `csv.writer`. Each call to `.writerow()` writes one row, given as a list. The first row we write is usually the header.

`newline=""` stops `csv.writer` from adding an extra blank line between rows on Windows.

**Be careful:** opening a file with `"w"` replaces anything that was already in it.

```python
with open("winners.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Year", "Title", "Author"])   # the header row
    for book in nominees:
        if book["Status"] in ["winner", "winner (tie)"]:
            writer.writerow([book["Year"], book["Title"], book["Author"]])

print("Saved winners.csv")
```

## 6.4 Finding the biggest value

To find the biggest value in a dictionary:
1. **Before** the loop, start with a value smaller than anything we'll see (`0` for a count).
2. **Inside** the loop, if the current value is bigger than the biggest so far, remember it, and remember its key too.

Because we use `>` and not `>=`, if two keys are tied, we keep the **first** one we found.

```python
most_nominations = 0
most_nominated = ""

for author in nomination_counts:
    if nomination_counts[author] > most_nominations:
        most_nominations = nomination_counts[author]
        most_nominated = author

print(f"{most_nominated} was nominated {most_nominations} times")
```

## 6.5 `range()` with a start, stop, and step

`range(start, stop, step)` gives us a sequence of numbers to loop over:
- It begins at `start`.
- It goes **up to but not including** `stop`, just like a slice.
- It counts by `step`. A negative step counts down.

With only one argument, `range(stop)` starts at `0` and counts up by 1.

```python
for count in range(6, 2, -1):
    print(count)

print("---")

for line_number in range(3):
    print("Line", line_number)
```

## 6.6 Nested loops

A loop can go inside another loop. The **inner** loop runs all the way through once for **each** time through the **outer** loop.

In Week 6 we used this to list the authors from most to least nominated. For each count, from the biggest down, we loop over every author to find the authors with that count.

```python
for count in range(most_nominations, 3, -1):
    for author in nomination_counts:
        if nomination_counts[author] == count:
            print(f"{count}  {author}")
```

## 6.7 Totals, averages, and percentages

These use the accumulator pattern from Week 3:
- An **average** is the total divided by how many things there are.
- A **percentage** is the part divided by the whole, times 100.

In an f-string, `{average:.2f}` shows a number rounded to 2 decimal places.

```python
total_nominations = 0
for author in nomination_counts:
    total_nominations = total_nominations + nomination_counts[author]

number_of_authors = len(nomination_counts)
average = total_nominations / number_of_authors
print(f"On average, an author is nominated {average:.2f} times")

nominated_once = 0
for author in nomination_counts:
    if nomination_counts[author] == 1:
        nominated_once = nominated_once + 1

percent = nominated_once / number_of_authors * 100
print(f"{percent:.1f}% of authors were nominated only once")
```

## 6.8 Fixing values with a dictionary of corrections

When the same thing is written in different ways (like `"Ontario"` and `"ON"`), we can fix it with a dictionary. The keys are the values we want to change, and each value is what to change it to. If a value isn't a key in the dictionary, we leave it as it is.

```python
corrections = {
    "Ontario": "ON",
    "Alberta": "AB",
    "British Columbia": "BC",
}


def clean_region(region):
    """ Return the corrected version of region, or region unchanged if it
        doesn't need correcting.
    """
    if region in corrections:
        return corrections[region]
    return region


print(clean_region("Ontario"))
print(clean_region("ON"))
print(clean_region("Nigeria"))
```

# Week 7

Week 7 put the tools from Weeks 5 and 6 together to ask how Canadian the Giller Prize is. These are the patterns that came up along the way.

## 7.1 Look before you count

Before counting, it's worth printing the values that might not fit our method. Remember that `in` also works on strings: it checks whether one string appears inside another. Here we find the authors who live in more than one place.

```python
for book in nominees:
    if " and " in book["Residence"]:
        print(f"{book['Author']}: {book['Residence']}")
```

## 7.2 A lookup dictionary: one value per author

Some authors were nominated more than once, so they appear in several rows. To get **one** value per author, we build a dictionary with authors as the keys. If an author appears again, assigning to the same key just replaces the value, so each author is only stored once.

We skip empty values with `!= ""`.

```python
birthplaces = {}

for book in nominees:
    born = book["Born"]
    if born != "":
        birthplaces[book["Author"]] = born

print(len(birthplaces), "authors")
print(birthplaces["Shyam Selvadurai"])
```

## 7.3 Starting every count at 0

Sometimes we know the keys we want to count ahead of time. Then we can start every key at `0` with one loop, and count with a second loop. Because every key is already there, we don't need the `else` from the counting pattern. We only count the values that are keys.

The keys that never show up keep a count of `0`, so they appear in our results too.

```python
prairie_counts = {}

# Start every province at 0
for province in ["AB", "SK", "MB"]:
    prairie_counts[province] = 0

# Count the authors born in each one
for author in birthplaces:
    parts = birthplaces[author].split(",")
    region = parts[-1].strip()
    if region in prairie_counts:
        prairie_counts[region] = prairie_counts[region] + 1

print(prairie_counts)
```

## 8.1 Loading a table with pandas

**pandas** is a module for working with tables. We import it with the short name `pd`.

`pd.read_csv()` reads a whole CSV file in one line. It gives us a **DataFrame**: a table with named columns and numbered rows. `len()` tells us the number of rows, and `.head()` shows the first five.

Unlike the `csv` module, pandas turns columns that hold only numbers into numbers. So in pandas, `Year` holds integers, not strings.

```python
import pandas as pd

books = pd.read_csv("giller_nominees_clean.csv")

print(len(books), "books")
print(type(books))
books.head()
```

## 8.2 Choosing columns

- One column name in square brackets gives us that **column**, called a **Series**.
- A **list** of column names gives us a smaller **DataFrame**. Notice the two pairs of square brackets: the outer pair chooses columns, and the inner pair makes the list.

```python
print(books["Title"].head())

books = books[["Status", "Year", "Title", "Author"]]
books.head()
```

## 8.3 Counting with `.value_counts()`

`.value_counts()` counts how many times each value appears in a column, from most to least common. It does the same job as the counting pattern from Week 5, in one line.

```python
status_counts = books["Status"].value_counts()
status_counts
```

## 8.4 Looking up one value in a Series

The result of `.value_counts()` is a Series. Each value has a label on the left. To get one value, we put its label in square brackets, just as we look up a key in a dictionary.

Because pandas reads the years as integers, we look up a year **without** quotation marks.

```python
print(status_counts["winner"])

year_counts = books["Year"].value_counts()
print(year_counts[2011])
```

## 8.5 From counts to percentages

`.value_counts(normalize=True)` gives each value's **fraction** of the rows instead of its count. Multiplying by 100 turns the fractions into percentages, and `.round(1)` rounds each one to 1 decimal place.

We don't need a loop to multiply by 100: pandas does the calculation for every value in the Series.

```python
status_percent = books["Status"].value_counts(normalize=True) * 100
status_percent = status_percent.round(1)
status_percent
```

## 8.6 Filtering rows with `==` and `!=`

Filtering keeps only the rows we want. It takes two steps:
1. A comparison on a column, like `books["Year"] == 1996`, gives a Series of `True` and `False`, one for each row.
2. Putting that Series inside `books[...]` keeps only the rows where it is `True`.

`!=` means "not equal to", so it keeps the rows that **don't** match.

```python
in_1996 = books["Year"] == 1996
print(in_1996.head())

books_1996 = books[in_1996]
print(len(books_1996), "books nominated in 1996")

not_long_listed = books[books["Status"] != "long-listed"]
print(len(not_long_listed), "books that made it past the long list")
```

## 8.7 Filtering with `.isin()`

`.isin()` checks whether each value is in a list. Like `==`, it gives a Series of `True` and `False` that we can filter with. It's useful when more than one value counts as a match, like the two kinds of winner.

```python
winners = books[books["Status"].isin(["winner", "winner (tie)"])]

print(len(winners), "winning books")
winners.head()
```

## 8.8 Filtering a Series of counts

We can filter the result of `.value_counts()` in the same two steps. Here we keep only the authors who won at least twice.

```python
win_counts = winners["Author"].value_counts()

at_least_two = win_counts >= 2
win_counts[at_least_two]
```

## 8.9 Charts  (This will not be on the test)

pandas can draw a chart from a Series. `.plot.barh()` draws a horizontal bar chart. We used **matplotlib** to add a title and labels.

```python
import matplotlib.pyplot as plt

ax = status_counts.plot.barh(color="#2a78d6")
ax.invert_yaxis()   # put the biggest bar at the top
ax.set_title("Most Giller nominees are long-listed")
ax.set_xlabel("Number of books")
plt.show()
```
