## 2.2 Comparison operators, and `=` vs `==`

Comparison operators ask a question and evaluate to a Boolean (`True` or `False`):

`==` (equal to), `!=` (not equal to), `>`, `>=`, `<`, `<=`

**Be careful:** a single `=` *assigns* a value to a variable. A double `==` *asks whether* two values are equal.

````python
line_count = 14          # assignment: store 14 in line_count

print(line_count == 14)  # comparison: is line_count equal to 14?
print(line_count != 14)
print(line_count > 19)
   
print("True" == True)    # a string is not equal to a Boolean
````

## 3.1 Indexing strings

A string is a **sequence** of characters, and we can get one character by its **index** (position) using square brackets. Indexing starts at **0**, not 1. Negative indexes count from the end: `-1` is the last character.

| index | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| character | r | a | v | e | n |
| negative index | -5 | -4 | -3 | -2 | -1 |

Using an index past the end (like `word[5]` here) causes an `IndexError`.

````python
word = "raven"

print(word[0])
print(word[2])
print(word[-1])
````

## 3.2 String methods

A **method** is a function that belongs to a particular data type. It is written *after* the value, attached with a dot: `value.method(arguments)`.

String methods we've used:

| Method | What it does |
|---|---|
| `.lower()` | all lowercase |
| `.upper()` | all uppercase |
| `.title()` | first letter of each word capitalized |
| `.replace(old, new)` | replace every `old` with `new` (takes two arguments) |
| `.isdigit()` | `True` if every character is a digit |
| `.split()` | break a string into a list of words (split on whitespace) |

````python
line = "Do not go gentle into that good night"

print(line.lower())
print(line.upper())
print(line.title())
print(line.replace("good night", "foul day"))
print("AH1".isdigit())
print("1".isdigit())
print(line.split())
````

## 3.3 Methods don't change the original string

Calling a string method gives back a **new** string; the original variable is unchanged. To keep the result, assign it to a variable (either a new one or the same one).

````python
opening = "It was the best of times, it was the worst of times"

opening.upper()
print(opening)            # still unchanged

opening = opening.upper() # reassign to keep the result
print(opening)
````

## 3.4 Indexing lists

Lists are also sequences, so indexing works exactly the same way as for strings: start at `0`, and use negative numbers to count from the end.

Here is the CMU Pronouncing Dictionary pronunciation of *elephant*, as a list of phonemes:

````python
elephant_phonemes = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

print(elephant_phonemes[0])
print(elephant_phonemes[-1])

# Indexes can be combined: the last character of the first phoneme
print(elephant_phonemes[0][-1])
````

## 3.5 `len()`

The function `len()` gives the length of a sequence: the number of characters in a string, or the number of items in a list.

````python
print(len("elephant"))          # number of characters
print(len(elephant_phonemes))   # number of phonemes in the list
````

## 3.6 `for` loops

A `for` loop repeats a block of code once for each item in a sequence. Each time through the loop, the **loop variable** (here `poet`) holds the next item.

The line starting the loop ends with a colon `:`, and the **body** of the loop is **indented**. Code that is not indented runs once, after the loop ends.

````python
poets = ["Montfort", "Lilian-Bertram", "Thomas", "Diaz", "Shakespeare"]

for poet in poets:
    print(f"A poem by {poet} was discussed in class.")

print("The loop is finished.")
````

## 3.7 The accumulator pattern

To count or build something up across a loop:
1. **Before** the loop, create a variable with a starting value (`0` for a count, `""` for an empty string).
2. **Inside** the loop, update it (for example, `count = count + 1`).
3. **After** the loop, use the result.

````python
line_words = ["shall", "i", "compare", "thee", "to", "a", "summer's", "day"]

letter_count = 0
for word in line_words:
    letter_count = letter_count + len(word)

print(f"The line has {letter_count} characters, not counting spaces.")
````

## 3.8 `if` / `elif` / `else`

An `if` statement runs its indented body only when its condition is `True`.
- `elif` ("else if") adds another condition to check.
- `else` runs when none of the conditions above it were `True`.

Conditions are checked **in order from top to bottom**, and only the **first** matching branch runs.

````python
word = "temperate"

if len(word) <= 3:
    print(f"'{word}' is a short word")
elif len(word) <= 6:
    print(f"'{word}' is a medium word")
else:
    print(f"'{word}' is a long word")
````

### Putting Week 3 together: counting syllables

In the CMU dictionary, every vowel phoneme ends in a digit marking its stress. So the number of syllables is the number of phonemes whose last character is a digit. This combines indexing, a string method, a loop, an `if` statement, and the accumulator pattern.

````python
word = "elephant"
word_p = ['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']

stress_pattern = ""
for phoneme in word_p:
    last_character = phoneme[-1]
    if last_character.isdigit():
        stress_pattern = stress_pattern + last_character

print(f"{word} has {len(stress_pattern)} syllables and stress pattern {stress_pattern}")
````

## 4.1 Slicing lists and strings

A **slice** takes a piece of a sequence using `[start:stop]`. It includes `start` and goes **up to but not including** `stop`.
- Leave out `start` to begin at the beginning: `[:3]`
- Leave out `stop` to go to the end: `[2:]`
- Negative numbers count from the end: `[:-1]` is everything except the last item.

Slicing works the same way on lists and strings. A slice of a list is a list; a slice of a string is a string.

````python
words = ["let", "me", "not", "to", "the", "marriage", "of", "true", "minds"]

print(words[0:3])
print(words[:3])
print(words[5:])
print(words[:-1])
print(words[2:5])

word = "marriage"
print(word[:4])
print(word[-3:])
````

## 4.2 Defining your own functions

We can write our own function with `def`, so we don't have to repeat the same code over and over:

```
def function_name(parameter):
    ''' Description of what the function does. '''
    body of the function (indented)
    return result
```

- The **parameter** is a variable that holds whatever argument is passed in when the function is called.
- The triple-quoted **docstring** right after the `def` line describes the function.
- Defining a function doesn't run it. The body only runs when the function is **called**.

````python
def get_stress_pattern(phonemes):
    ''' Return the stress pattern, as a string, of the given list of phonemes.

        Example: get_stress_pattern(['AH0', 'W', 'EY1', 'K']) returns "01"
    '''
    stress_pattern = ""
    for phoneme in phonemes:
        last_character = phoneme[-1]
        if last_character.isdigit():
            stress_pattern = stress_pattern + last_character
    return stress_pattern


print(get_stress_pattern(['AH0', 'W', 'EY1', 'K']))                 # awake
print(get_stress_pattern(['EH1', 'L', 'AH0', 'F', 'AH0', 'N', 'T']))  # elephant
````

## 4.3 `return` and `None`

`return` sends a value back to wherever the function was called, and **ends** the function immediately. A function can have more than one `return`, but only one of them runs each time.

`None` is a special Python value meaning "no value". It's useful to return when there is no valid answer. (A function with no `return` statement also gives back `None`.)

**`return` is not the same as `print()`:** `print()` only displays a value; `return` hands the value back so it can be stored in a variable and used later.

````python
def first_word(line):
    ''' Return the first word of line, or None if line has no words. '''
    words = line.split()
    if len(words) == 0:
        return None
    return words[0]


opening = first_word("Whose woods these are I think I know")
print(opening)
print(first_word(""))
````

## 4.4 `in` and `not in`

`a in b` evaluates to `True` if `a` is an item in the list `b` (or, for strings, if `a` appears inside the string `b`). `not in` gives the opposite. These are most often used in `if` statements.

(This is a different job from the `in` in a `for` loop.)

````python
stop_words = ["the", "a", "an", "of", "to", "and"]

print("the" in stop_words)
print("love" in stop_words)
print("love" not in stop_words)
print("ear" in "heart")       # works on strings too
````

## 4.5 Building a list with `.append()`

`.append()` is a **list** method that adds one item to the end of a list. Unlike string methods, it **changes the original list**.

A common pattern: start with an empty list `[]`, loop, and append the items you want to keep. Combined with `not in`, this builds a list of unique words.

````python
sonnet_words = ["love", "is", "not", "love", "which", "alters",
                "when", "it", "alteration", "finds"]

word_bank = []
for word in sonnet_words:
    if word not in word_bank:
        word_bank.append(word)

print(word_bank)
print(f"{len(sonnet_words)} words, {len(word_bank)} unique")
````

## 4.6 `.split()` and `.join()`

These two string methods go in opposite directions:
- `line.split()` turns a **string into a list** of words. When no argument is given to split, it will split on whitespace (blanks, newline characters, and tabs)
- `separator.join(list_of_strings)` turns a **list into a string**, putting `separator` between each item.

The `.join()` syntax reads backwards from what you might expect: the separator comes first.

````python
# Remember to try to predict what the output will be

line = "the woods are lovely dark and deep"

words = line.split()
print(words)

print(" ".join(words))
print(" / ".join(words))
print(" ".join(words[:3]))
````

## 4.7 Combining conditions with `and` and `not`

- `condition1 and condition2` is `True` only if **both** are `True`.
- `not condition` flips `True` to `False` and vice versa.

Empty strings, empty lists, and `None` all count as false in an `if` statement, so `if not some_value:` is a quick way to check for "nothing there".

````python
# Example 1
stress_word1 = "0"
stress_word2 = "1"

if stress_word1 == "0" and stress_word2 == "1":
    print("This pair of one-syllable words makes an iamb.")

# Example 2
pronunciation = None
if not pronunciation:
    print("No pronunciation found.")
````
