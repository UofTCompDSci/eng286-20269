### Task 3: The `in` Operator

Here is a short list of common *stop words*.  Stop words are words so frequent that they don't tell us much about a text on their own:

```python

```

1. Write an expression that checks whether `"love"` is in `stop_words`, and store the result of that expression (`True` or `False`) in `is_stop_word`.

3. Set `test_word2 = "thee"` and use `not in` to add `"thee"` to `stop_words`, but only if it isn't already there.

````python
# Solution
# Write your code here
stop_words = ['the', 'a', 'an', 'of', 'to', 'and']

# 1. Is "love" a stop word?
is_stop_word = "love" in stop_words

# Check your work
print(is_stop_word)

# 2. if/else with test_word1 = "of"
test_word1 = "of"

if test_word1 in stop_words:
    print(f"{test_word1} is a stop word")
else: 
    print(f"{test_word1} is not a stop word")


# 3. Add test_word2 to stop_words if it isn't already there.
test_word2 = "thee"

if test_word2 not in stop_words:
    stop_words.append(test_word2)

print(stop_words)
````

````python
# Solution
# Write your code here
word_list = ["why", "lov'st", "thou", "that", "which", "thou", "receiv'st", "not", "gladly"]

apostrophe_count = 0

for word in word_list:
    if "'" in word:
        apostrophe_count = apostrophe_count + 1

# Check your work
print(apostrophe_count)
````

## Extra Practice 1

Take the `if`/`elif`/`else` chain from Task 5 and apply it to every word in a list, keeping three separate running counts: `vowel_count`, `y_count`, and `consonant_count`. The goal is to do this in a single loop.

```python
word_bank = ['elephant', 'young', 'iamb', 'sonnet', 'yellow', 'orange', 'stress',
             'unstressed', 'why', 'gladly']
```

````python
# Solution
# Write your code here
word_bank = ['elephant', 'young', 'iamb', 'sonnet', 'yellow', 'orange', 'stress', 'unstressed', 'why', 'gladly']
vowels = "aeiou"

vowel_count = 0
y_count = 0
consonant_count = 0

for word in word_bank:
    if word[0] in vowels:
        vowel_count = vowel_count + 1
    elif word[0] == 'y':
        y_count = y_count + 1
    else:
        consonant_count = consonant_count + 1

# Check your work
print(vowel_count)
print(y_count)
print(consonant_count)
````
