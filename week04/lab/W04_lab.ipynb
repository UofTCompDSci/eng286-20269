In lecture we found a word's stress pattern by looking at its pronunciation from the CMU Pronouncing Dictionary. Every vowel sound ends with a digit: `1` for a stressed syllable, `0` for an unstressed one (and `2` for secondary stress).

The code cell below is meant to go through the pronunciation of the word "extremity" (stored in `word_p`) and create two variables:
- `stress_pattern`: a string of the stress digits in order. For "extremity" this should be `"0100"` (ex-**TREM**-i-ty).
- `syllable_count`: the number of syllables, which is the number of phonemes that end in a digit.

But there are **three** mistakes in the code! Fix them directly in the cell below (don't create a new cell) so that the code runs without errors and produces the variables described above.

```python
word = "extremity"
word_p = ['IH0', 'K', 'S', 'T', 'R', 'EH1', 'M', 'AH0', 'T', 'IY0']

stress_pattern = ""
syllable_count = 0

for phoneme in word_p
    last_character = phoneme[0]
    if last_character.isdigit():
        stress_pattern = stress_pattern + last_character
syllable_count = syllable_count + 1

print(f"The stress pattern of {word} is {stress_pattern}")
print(f"{word} has {syllable_count} syllables")
```
